# ELM-FUN N/P limitation indices — calculation & evaluation
For Tom Bytnerowicz. Computed from the **corrected** runs (3 sites x 3 FUN schemes, annual 1850-2100),
benchmarked against **Braghiere et al. 2022 (JAMES)** and **Menge et al. 2026 (GBC)**.
Full write-up: `docs/LIMITATION_INDICES_EVALUATION.md`.

**Bottom line:** the indices make sense as *relative / dynamic* measures, with two caveats on *absolute* magnitude —
(A) ELM-FUN realizes the C price of N/fixation very low (~0.1-0.3 gC/gN vs obs ~5), so cost-based limitation is
meaningful relatively but not as absolute intensity; (B) leaf C:N / N:P are prescribed (flat) so they cannot serve
as a warming-response index — the dynamic signal is in retranslocation.

## Verdict (see docs/LIMITATION_INDICES_EVALUATION.md for full detail + tables)
1. **Tom's N/P split** — valid & internally consistent; P-acquisition dominates N everywhere (P-limited forests).
2. **Fixation confound** — NPP_NFIX is 7-8% of NPP_NUPTAKE at Manaus/Harvard (N-fixing sites). Use NPP_NACTIVE for a fixation-free N index.
3. **CAVEAT A** — C price of N/fixation realized ~0.1-0.3 gC/gN (cross-checked), ~40x below obs. Fluxes realistic, C investment tiny (s_fix tuning). Cost-based limitation is relative, not absolute.
4. **CAVEAT B** — leaf C:N/N:P fixed => null warming response by construction. Portable across models but flag it. Dynamic analog = retranslocation N:P.
5. **Warming x fixation** — matches the protocol: Houlton suppresses warm-site fixation most, Bytnerowicz sustains.
6. **Model's NUPTAKE_NPP_FRACTION != Tom's proxy** (denominator=availc); ~5x lower; not interchangeable.

**Recommended for Tom:** `NPP_NUPTAKE/NPP_PUPTAKE` (N-vs-P balance), `NPP_NACTIVE/(...)` (fixation-free N),
fixation share + `NPP_NFIX/NPP` (warming x fixation, w/ Caveat A vs Menge 2026), leaf N:P (portable, w/ Caveat B).
